```python
Mark_1=int(input("Enter the mark 1:"))
Mark_2=int(input("Enter the mark 2:"))
Mark_3=int(input("Enter the mark 3:"))
average=(Mark_1+Mark_2+Mark_3)/3
print("Average Percentage:",average)
if average>=90:
    print("Grade A")
elif average>=75:
    print("Grade B")
elif average>=50:
    print("Grade C")
else:
    print("Fail")
```

```python
n = int(input("Enter N: "))

i = 1
sum = 0

while i <= n:
    sum = sum + i
    i = i + 1

average = sum / n

print("Sum =", sum)
print("Average =", average)
```

```python
num = int(input("Enter an integer: "))

if num < 2:
    print("Not a Prime Number")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not a Prime Number")
            break
    else:
        print("Prime Number")
```

```python
num = int(input("Enter an integer: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number:", reverse)

if original == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
    
```

```python
http_code=int(input("Enter a http code:"))
match http_code:
    case 200:
        print("OK Successful")
    case 400:
        print("Bad request")
    case 404:
        print("Not found")
    case 500:
        print("Internal server error")
    case _:
        print("Invalid Status Code")
        
```
